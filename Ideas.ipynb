Has agricultural water use become decoupled from agricultural production in France?
Hypothesis:
1. Agricultural water withdrawal intensity has decreased over time in at least some French regions.
2. Improvements in water withdrawal intensity do not necessarily translate into reductions in total agricultural water withdrawals.
3. The relationship between agricultural production and water withdrawals varies substantially across French regions.
